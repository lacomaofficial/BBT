```python
import os
from bbtransformer import run_tuning_workflow
```

```python
TUNING_CONFIG = {
    'epochs': 5000,
    'patience': 100,
    'early_stop_metric': 'f1',
    'use_focal_loss': False,

    # Optimizer: narrow range covering both best trials
    'lr_range': (2.4e-5, 2.5e-5),                 
    'weight_decay_range': (1.13e-6, 1.79e-6),     

    # Architecture: fixed based on both models
    'embed_dim': [512],
    'num_heads': [16],
    'num_layers_range': (7, 7),                   
    'n_kv_heads_options': [8],                 

    # Embeddings & patching: fixed across both models
    'embed_dim_age': [32],                       
    'embed_dim_ext': [16],
    'patch_size': [3],
    'patch_embed_ratio': [0.75],
    'temp_attn_hidden': [512],

    # Dropout ranges: min–max from the two trials
    'dropout_input_range': (0.17, 0.19),
    'dropout_patch_range': (0.15, 0.20),
    'dropout_attn_range': (0.15, 0.16),
    'dropout_ffn_range': (0.24, 0.27),
    'dropout_classifier_range': (0.04, 0.06),
    'dropout_temporal_range': (0.16, 0.17),

    # Regularization
    'stochastic_depth_rate_range': (0.10, 0.12),

    # Evaluation
    'metric_threshold': 0.60,
    'metric_weights': {
        'f1': 0.20,
        'roc_auc': 0.20,
        'accuracy': 0.20,
        'precision': 0.20,
        'recall': 0.20
    },

    # Tuning strategy
    'n_startup_trials': 3,
    'n_warmup_steps': 0,
    'interval_steps': 1,
    'base_seed': 42,
}
```

```python
# === DISORDER PATHS ===
CHRT_BASE_DIR = '/mnt/movement/users/jaizor/xtra/data/fmri/chrt'

def get_chrt_paths(disorder):
    return (
        os.path.join(CHRT_BASE_DIR, f'fmri_{disorder}.npz'),
        os.path.join(CHRT_BASE_DIR, f'pheno_{disorder}.csv')
    )

TARGET = 'NervousSystem_Multiple_Sclerosis_Other_Demyelinating'  

DATA_PATH, PHENO_PATH = get_chrt_paths(TARGET)
```

```python
# For Parkinson’s
run_tuning_workflow(
    target_name=TARGET,
    data_path=DATA_PATH,
    pheno_path=PHENO_PATH,
    tuning_config=TUNING_CONFIG,
    n_trials=50
)
```
